```python
import spacy
from spacy.lang.en.stop_words import STOP_WORDS
from string import punctuation
```

```python
stopwords = list(STOP_WORDS)
stopwords
```

```python
len(stopwords)
```

```python
nlp = spacy.load('en_core_web_sm')
```

```python
text
```

```python
doc = nlp(text)
doc
```

```python
# Lets get token from the text
tokens = [token.text for token in doc]
print(tokens)
```

```python
doc
```

```python
# w=We have to calculate the frequency of each and every token, how many times is repition
word_frequencies = {}

for word in doc:
  if word.text.lower() not in stopwords:
    if word.text.lower() not in punctuation:
      if word.text not in word_frequencies.keys():
         word_frequencies[word.text] = 1
      else:
        word_frequencies[word.text] += 1
```

```python
word_frequencies
```

```python
len(word_frequencies)
```

```python
max_frequency = max(word_frequencies.values())
max_frequency
```

```python
# to get normalized/weighted frequencies you should divide all the frequency with 11

for word in word_frequencies.keys():
  word_frequencies[word] = word_frequencies[word]/max_frequency
```

```python
word_frequencies
```

```python
sentence_tokens = [sent for sent in doc.sents]
sentence_tokens
```

```python
len(sentence_tokens)
```

```python
# We are going to calculate the sentence acore, to calculate the sentence core

sentence_score = {}

for sent in sentence_tokens:
  for word in sent:
    if word.text.lower() in word_frequencies.keys():
      if sent not in sentence_score.keys():
        sentence_score[sent] = word_frequencies[word.text.lower()]
      else:
        sentence_score[sent] += word_frequencies[word.text.lower()]
```

```python
sentence_score
```

```python
len(sentence_score)
```

```python
# Lets say our case study was 30 % sentence with maximum scores

from heapq import nlargest
```

```python
select_length = int(len(sentence_tokens)*0.4)
select_length
```

```python
# We have to select maximum 40% sentences out of all sentences

summary = nlargest(select_length, sentence_score, key = sentence_score.get)
summary
```

```python
# If i have to combine these top 3 sentences then
final_summary = [word.text for word in summary]
final_summary
```
